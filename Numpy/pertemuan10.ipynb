Exercise 1 — Sorting Arrays (Mengurutkan Array)

```python
import numpy as np

# 1. Array 1D temperatures
temperatures = np.array([30, 27, 33, 29, 31, 28])

# Urutkan menaik (ascending) - np.sort TIDAK mengubah array asli
sorted_asc = np.sort(temperatures)
print("Menaik:", sorted_asc)

# Urutkan menurun (descending) - gunakan slicing [::-1] pada hasil sort
sorted_desc = np.sort(temperatures)[::-1]
print("Menurun:", sorted_desc)

# 2. Matriks rainfall
rainfall = np.array([[8, 3, 6], 
                     [2, 9, 5], 
                     [7, 1, 4]])

# Urutkan berdasarkan kolom (axis=0)
print("\nSorted axis=0 (per kolom):\n", np.sort(rainfall, axis=0))

# Urutkan berdasarkan baris (axis=1)
print("\nSorted axis=1 (per baris):\n", np.sort(rainfall, axis=1))
```

Exercise 2 — Filtering Arrays (Menyaring Array)

```python
scores = np.array([65, 80, 92, 74, 88, 95, 70])

# 1. Minimal 80 (Boolean Mask)
print(">= 80:", scores[scores >= 80])

# 2. Antara 70 dan 90 (termasuk keduanya). PENTING: Gunakan tanda kurung & operator '&'
print("70 - 90:", scores[(scores >= 70) & (scores <= 90)])

# 3. Fancy indexing untuk indeks 0, 3, dan 5
print("Indeks 0, 3, 5:", scores[[0, 3, 5]])

# 4. np.where() untuk menemukan INDEKS nilai > 90
print("Indeks > 90:", np.where(scores > 90))
```

Exercise 3 — Updating Arrays (Memperbarui Array)

```python
inventory = np.array([[10, -1], 
                      [15, 8], 
                      [12, -3]])

# 1. Salinan dan ganti nilai negatif jadi 0
inv_copy = inventory.copy()
inv_copy[inv_copy < 0] = 0
print("1. Negatif jadi 0:\n", inv_copy)

# 2. Tambah baris baru [[20, 5]]
# Perhatikan: argumen np.concatenate harus berupa tuple: (array1, array2)
inv_with_row = np.concatenate((inv_copy, [[20, 5]]), axis=0)
print("\n2. Setelah tambah baris:\n", inv_with_row)

# 3. Tambah kolom [[1], [2], [3], [4]]
inv_with_col = np.concatenate((inv_with_row, [[1], [2], [3], [4]]), axis=1)
print("\n3. Setelah tambah kolom:\n", inv_with_col)

# 4. Hapus baris pertama (index 0) dan kolom terakhir (index -1)
final_inv = np.delete(inv_with_col, 0, axis=0) # Hapus baris 0
final_inv = np.delete(final_inv, -1, axis=1)   # Hapus kolom terakhir
print("\n4. Hasil akhir:\n", final_inv)
print("Shape akhir:", final_inv.shape)
```

Exercise 4 — Array Aggregation (Agregasi Array)

```python
visitors = np.array([[20, 25, 30], 
                     [18, 22, 28], 
                     [24, 27, 32]])

# 1. Total dan rata-rata seluruh pengamatan
print("Total:", np.sum(visitors))
print("Rata-rata:", np.mean(visitors))

# 2. Total per kolom (axis=0) dan rata-rata per baris (axis=1)
print("Total per kolom:", np.sum(visitors, axis=0))
print("Rata-rata per baris:", np.mean(visitors, axis=1))

# 3. Jumlah kumulatif pada setiap baris
print("Cumsum per baris:\n", np.cumsum(visitors, axis=1))

# 4. Rata-rata per baris dengan keepdims=True
mean_keepdims = np.mean(visitors, axis=1, keepdims=True)
print("\nMean dengan keepdims:\n", mean_keepdims)
print("Shape-nya:", mean_keepdims.shape) # Akan menjadi (3, 1), bukan (3,)
```

Exercise 5 — Vectorized Operations (Operasi Vektorisasi)

```python
prices = np.array([10000, 15000, 20000, 25000])
quantities = np.array([2, 3, 1, 4])

# 1. Pendapatan tiap produk (element-wise multiplication)
revenue = prices * quantities
print("Pendapatan:", revenue)

# 2. Total pendapatan
print("Total Pendapatan:", np.sum(revenue))

# 3. Diskon 10% (harga menjadi 90%)
discounted_prices = prices * 0.9
print("Harga setelah diskon:", discounted_prices)

# 4. Boolean mask untuk pendapatan >= 50000
print("Mask pendapatan >= 50000:", revenue >= 50000)
```

Exercise 6 — Integrated Statistical Practice

```python
student_results = np.array([[80, 75, 90],
                            [65, 70, 72],
                            [92, 88, 95],
                            [78, 85, 82]])

# 1. Rata-rata nilai masing-masing mahasiswa (per baris)
student_means = np.mean(student_results, axis=1)
print("Rata-rata per mahasiswa:", student_means)

# 2. Pilih baris mahasiswa yang rata-ratanya minimal 80
high_achievers = student_results[student_means >= 80]
print("\nMahasiswa dengan rata-rata >= 80:\n", high_achievers)

# 3. Posisi mahasiswa dengan nilai rata-rata tertinggi memakai np.argsort()
# np.argsort mengurutkan indeks dari terkecil ke terbesar. [-1] mengambil yang terakhir (terbesar)
highest_pos = np.argsort(student_means)[-1]
print("\nPosisi mahasiswa terbaik (indeks):", highest_pos)

# 4. Tambah 5 poin, batasi maksimal 100 menggunakan np.where()
# Sintaks: np.where(kondisi, nilai_jika_true, nilai_jika_false)
updated_results = np.where(student_results + 5 > 100, 100, student_results + 5)
print("\nNilai setelah ditambah 5 (maks 100):\n", updated_results)

# 5. Rata-rata per ujian (per kolom) pada data yang telah diperbarui
final_exam_means = np.mean(updated_results, axis=0)
print("\nRata-rata per ujian:", final_exam_means)
```
